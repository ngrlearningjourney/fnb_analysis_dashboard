```python
import pandas as pd
```

Loading Dataset

```python
df = pd.read_csv('/content/synthetic_fnb_order_items_bridge_1000.csv')
df.head()
```

Understanding the Dataset Shape and Data

Based on the code result, it is expected to contain **1,000 records**. However, there are **10 missing values** in the **Product Name** column and **10 missing values** in the **Rating** column.

```python
df.info()
```

`df["rating"].fillna(df['rating'].median(), inplace = True)` is a code to fill missing value on rating column with median value from rating column

```python
df["rating"].fillna(df['rating'].median(), inplace = True)
```

`df = df.dropna()` is a code to drop any rows that contain missing value inside the dataframe. This action taken because filling product name missing value is not possible

```python
df = df.dropna()
```

The `df.describe()` function helps users understand the numerical columns in the dataset by providing several descriptive statistics. Based on the results, we noticed that there are incorrect values in the **Price** column. Specifically, a few rows contain a value of **-3** in the **Price** column.

```python
df.describe()
```

`df['price'].replace(to_replace=-3,value=3, inplace=True)` is a code that replace -3 incorrect value on price column into 3

```python
df['price'].replace(to_replace=-3,value=3, inplace=True)
```

`del df['line_total']` and `df['line_total'] = df['price'] * df['quantity']` are two lines of code. The first line deletes the **line_total** column from the dataset. The second line creates a new **line_total** column and fills it with the result of multiplying the **price** and **quantity** columns.

```python
del df['line_total']
df['line_total'] = df['price'] * df['quantity']
```

`df["month"] = df['order_date'].str[5:7]` is a code that make one new column called month and fill that new column with order date data that have been substringed so it only take the month of every date inside the order date column

```python
df["month"] = df['order_date'].str[5:7]
```

`df['month'].replace(to_replace='/2',value='02',inplace=True)` this line of code replace value "/2" to "02" on month column

```python
df['month'].replace(to_replace='/2',value='02',inplace=True)
```

`df['region'].replace(to_replace='Unknown',value=df['region'].mode()[0],inplace=True)` is a line of code that replace value "unkown" with the mode of region column

```python
df['region'].replace(to_replace='Unknown',value=df['region'].mode()[0],inplace=True)
```
